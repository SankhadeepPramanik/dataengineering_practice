# Date Functions in Databricks SQL

**Goal:** Demonstrate key Databricks SQL date functions — `date_part`, `dayofyear`, `dateadd`, and `INTERVAL` arithmetic — and use them to solve a practical problem: finding the next occurrence of a specific weekday (Sunday).

```python
%sql
-- date_part(field, expr) — extracts a single field from a date/timestamp/interval
--   field can be: year, quarter, month, week, day, dayofweek,
--                hour, minute, second, epoch
--   (dayofyear is not a valid date_part field; use dayofyear(expr) instead)
SELECT
  date_part('year',      current_date()) AS yr,
  date_part('month',     current_date()) AS mon,
  date_part('dayofweek', current_date()) AS weekday,   -- 1 = Sunday … 7 = Saturday
  dayofyear(current_date())                AS doy;

-- dateadd(unit, value, expr) — adds/subtracts a time unit to a date/timestamp
--   unit can be: year, quarter, month, week, day, hour, minute, second,
--               millisecond, microsecond
--   value is an integer (negative to subtract)
SELECT
  dateadd(day,    7,  current_date())     AS plus_7_days,
  dateadd(day,   -1,  current_date())     AS yesterday,
  dateadd(month,  3,  current_date())     AS plus_3_months,
  dateadd(year,  -1,  current_date())     AS one_year_ago,
  dateadd(hour,   2,  current_timestamp()) AS plus_2_hours;

-- Common shorthand: you can also use +/- INTERVAL syntax
SELECT
  current_date() + INTERVAL 7 DAYS  AS plus_7_days_alt,
  current_date() - INTERVAL 1 DAY   AS yesterday_alt;
```

```python
%sql
-- Find the next occurrence of Sunday from today
-- Logic: dayofweek returns 1=Sunday … 7=Saturday
--   If today is Sunday (1), next Sunday = today + 7 days
--   If today is Wednesday (4), next Sunday = today + (8-4) = today + 4 days
--   General formula: dateadd(day, 8 - dayofweek(today), today)
select dateadd(day, 8-date_part('dayofweek',current_date()), current_date())
```
